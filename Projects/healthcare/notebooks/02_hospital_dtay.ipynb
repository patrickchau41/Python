### 2. Does a more intense hospital stay predit 30-day readmission?

```python
import pandas as pd
 
# Load the cleaned data made in notebook 01
df = pd.read_csv('../datasets/diabetic_clean.csv')
 
# Checkpoint: expect 69,982 rows and 50 columns
print(df.shape)
```

```python
# ---------- 1) Length of stay (days in the hospital) ----------
stay_group = pd.cut(df['time_in_hospital'], bins=[0, 2, 4, 7, 14],
                    labels=['1-2 days', '3-4 days', '5-7 days', '8-14 days'])
table_stay = df.groupby(stay_group, observed=True)['readmitted_30d'].agg(['count', 'mean'])
table_stay['mean'] = (table_stay['mean'] * 100).round(2)   
print('LENGTH OF STAY: patients and 30-day readmission %')
print(table_stay)
```

```python
# ---------- 2) Number of diagnoses recorded for the visit ----------
diag_group = pd.cut(df['number_diagnoses'], bins=[0, 4, 6, 8, 16],
                    labels=['1-4', '5-6', '7-8', '9-16'])
table_diag = df.groupby(diag_group, observed=True)['readmitted_30d'].agg(['count', 'mean'])
table_diag['mean'] = (table_diag['mean'] * 100).round(2)
print('DIAGNOSES: patients and 30-day readmission %')
print(table_diag)
```

```python
# ---------- 3) Number of different medications given ----------
# The top edge of 100 is just a safe upper limit so the largest values land in "21+"
med_group = pd.cut(df['num_medications'], bins=[0, 10, 15, 20, 100],
                   labels=['1-10', '11-15', '16-20', '21+'])
table_med = df.groupby(med_group, observed=True)['readmitted_30d'].agg(['count', 'mean'])
table_med['mean'] = (table_med['mean'] * 100).round(2)
print('MEDICATIONS: patients and 30-day readmission %')
print(table_med)
```

```python
# ---------- 4) Number of procedures (non-lab, e.g. surgery) ----------

proc_group = pd.cut(df['num_procedures'], bins=[-1, 0, 1, 2, 10],
                    labels=['0', '1', '2', '3+'])
table_proc = df.groupby(proc_group, observed=True)['readmitted_30d'].agg(['count', 'mean'])
table_proc['mean'] = (table_proc['mean'] * 100).round(2)
print('PROCEDURES: patients and 30-day readmission % (no clear pattern)')
print(table_proc)
```

```python
# ---------- 5) Number of lab tests ----------
lab_group = pd.cut(df['num_lab_procedures'], bins=[0, 30, 45, 60, 140],
                   labels=['1-30', '31-45', '46-60', '61+'])
table_lab = df.groupby(lab_group, observed=True)['readmitted_30d'].agg(['count', 'mean'])
table_lab['mean'] = (table_lab['mean'] * 100).round(2)
print('LAB TESTS: patients and 30-day readmission %')
print(table_lab)
```
