### 4. Are patient demographics linked to 30-day readmission?

```python
import pandas as pd

df = pd.read_csv('../datasets/diabetic_clean.csv', low_memory=False)
print(df.shape)   # should be (69982, 50)
```

```python
# Average of readmitted_30d = share of patients readmitted within 30 days
table_age = df.groupby('age')['readmitted_30d'].agg(['count', 'mean'])
table_age['mean'] = (table_age['mean'] * 100).round(2)
table_age.columns = ['patients', 'readmit_rate_%']
table_age
```

```python
table_gender = df.groupby('gender')['readmitted_30d'].agg(['count', 'mean'])
table_gender['mean'] = (table_gender['mean'] * 100).round(2)
table_gender.columns = ['patients', 'readmit_rate_%']
table_gender
```

```python
table_race = df.groupby('race')['readmitted_30d'].agg(['count', 'mean'])
table_race['mean'] = (table_race['mean'] * 100).round(2)
table_race.columns = ['patients', 'readmit_rate_%']
table_race.sort_values('patients', ascending=False)
```

## What it shows
- Age: readmission rises from about 7% for patients under 60 to about 10.8% for ages 80-90. The youngest groups are very small, so ignore them.
- Gender: Female 9.04% vs Male 8.91%, basically no difference.
- Race: Caucasian 9.19% vs African American 8.67%, a small gap. Smaller groups (Asian, Other, Hispanic) have few patients, so their rates are less reliable.
- Takeaway: age matters a bit, gender and race barely do. Prior hospital stays and discharge destination (Q1 and Q3) matter far more.
